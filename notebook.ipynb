```python
import statistics

ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

# a
print(len(ages))

# b
print(sum(ages))

# c
mean_age = sum(ages) / len(ages)
print(round(mean_age, 1))

# d
print(min(ages))
print(max(ages))

# e
median_age = statistics.median(ages)
print(median_age)
print(mean_age != median_age)
```

```python
import pandas as pd

df = pd.read_excel("50_startups_messy.xlsx")

# 01
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())

# 02
df = df.drop_duplicates()
df["State"] = df["State"].str.strip().str.title()
df["Marketing Spend"] = pd.to_numeric(df["Marketing Spend"], errors="coerce")

# 03
simple = df.sample(n=20, random_state=7)

stratified = df.groupby("State", group_keys=False).apply(
    lambda g: g.sample(frac=0.2))

# 04
print(df[["R&D Spend", "Marketing Spend"]].mean())
print(simple[["R&D Spend", "Marketing Spend"]].mean())
print(stratified[["R&D Spend", "Marketing Spend"]].mean())
```
